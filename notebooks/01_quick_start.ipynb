# 本notebook演示AdaPrune的基本用法

# %%
import sys
sys.path. insert(0, '. .')

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn. model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

from adaprune import AdaPruneGBDT

# %% [markdown]
# ## 1. 生成示例数据

# %%
X, y = make_classification(
    n_samples=2000,
    n_features=20,
    n_informative=15,
    n_redundant=3,
    flip_y=0.1,
    random_state=42
)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print(f"Training set:  {X_train. shape}")
print(f"Test set:  {X_test. shape}")

# %% [markdown]
# ## 2. 训练AdaPrune模型

# %%
model = AdaPruneGBDT(
    n_estimators=100,
    learning_rate=0.1,
    strategy='hybrid',
    adaptation_frequency=5,
    early_stopping_rounds=20,
    random_state=42,
    verbose=1
)

model.fit(X_train, y_train)

# %% [markdown]
# ## 3. 评估模型

# %%
y_pred = model. predict(X_test)
print("Classification Report:")
print(classification_report(y_test, y_pred))
print(f"\nTest Accuracy: {accuracy_score(y_test, y_pred):.4f}")

# %% [markdown]
# ## 4. 可视化自适应过程

# %%
fig = model.plot_adaptation_history(figsize=(14, 10))

# %% [markdown]
# ## 5. 对比不同策略

# %%
strategies = ['gap_based', 'trend_based', 'data_aware', 'hybrid']
results = {}

for strategy in strategies:
    m = AdaPruneGBDT(n_estimators=100, strategy=strategy, verbose=0)
    m.fit(X_train, y_train)
    test_acc = accuracy_score(y_test, m.predict(X_test))
    results[strategy] = test_acc
    print(f"{strategy: 15s}: {test_acc:.4f}")
